```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("Data Cleaning 12.08.csv")
df = pd.DataFrame(data)
df.head()
```

```python
#rows and columns
df.shape
```

```python
#null values
df.isnull().sum()
```

```python
#duplicated values
df.duplicated().sum()
```

```python
# Replace infinity values (inf) with NaN (null val)
df = df.replace([np.inf, -np.inf], np.nan)

# Fix any negative values in error columns by making them positive
error_cols = [col for col in df.columns if 'error' in col]
df[error_cols] = df[error_cols].abs()
```

```python
#fix missing mass vals - combine mass and mass_sini
df['mass'] = df['mass'].fillna(df['mass_sini'])
df.head()
```

```python
df.isnull().sum()
```

```python
#remove remaining missing 
df = df.dropna(subset=['mass'])
df.isnull().sum()
```

```python
# Fix missing errors by fixing the empty values with 0 (otherwise calculations might break)
error_cols = ['mass_error_min', 'mass_error_max', 'mass_sini_error_min', 'mass_sini_error_max', 'radius_error_min']
for col in error_cols:
    df[col] = df[col].fillna(0)
```

```python
df.isnull().sum()
```

```python
# missing radius
df['radius'] = df['radius'].fillna(df['radius'].median())
df.isnull().sum()
```

```python
# drop the redundant mass_sini column
df = df.drop(columns=['mass_sini'])
df.isnull().sum()
```
